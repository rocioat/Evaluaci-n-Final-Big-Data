##**Etapa 1. Problema y pregunta de análisis**


**Contexto Institucional**

El Servicio de Vivienda y Urbanización (SERVIU) es un organismo público encargado de materializar las políticas habitacionales del Estado, gestionando y supervisando la ejecución de obras de viviendas y parques urbanos. Dentro de sus funciones se encuentran la administración de subsidios habitacionales, es decir, velar por el adecuado desarrollo y gestión de los proyectos en todas sus etapas, desde la asignación de subsidios hasta la recepción final de las obras, asegurando el cumplimiento de plazos, requisitos normativos y compromisos adquiridos con los beneficiarios, así como la coordinación con entidades patrocinantes y empresas constructoras.

En el caso de los proyectos habitacionales del Programa Decreto Supremo N° 27, son principalmente obras de mejoramiento como: equipamiento comunitario para sedes sociales y/o plazas, reparaciones estructurales, instalaciones sanitarias y eléctricas, mejoramiento de condominios, fachadas, techumbres, escaleras, entre otras. El seguimiento de estas constituye una tarea relevante para la gestión institucional, debido a que permite visualizar y controlar los diversos factores que podrían influir en plazos de ejecución.



**Problema**

Los proyectos habitacionales ejecutados por SERVIU cuentan con una programación física y financiera que establece plazos para la ejecución de obras y el uso de recursos públicos. Sin embargo, la duración efectiva de los proyectos puede variar considerablemente entre una iniciativa y otra, incluso cuando pertenecen al mismo programa habitacional.

En el caso de los proyectos DS27, esta variabilidad puede estar asociada a múltiples características propias de cada proyecto, tales como el monto de inversión comprometido, la tipología de intervención, el número de unidades involucradas o la situación de avance de la obra. Como resultado, algunos proyectos logran ejecutarse dentro de los plazos previstos, mientras que otros presentan extensiones importantes en sus tiempos de ejecución.

Desde una perspectiva de gestión, conocer cuáles de estos factores tienen una mayor influencia sobre la duración de las obras resulta relevante para mejorar la planificación, la asignación de recursos y el seguimiento de los proyectos. Asimismo, permite generar evidencia que apoye la toma de decisiones y contribuya a una gestión más eficiente de los programas habitacionales.

En este contexto, surge la necesidad de analizar estadísticamente la relación entre la duración de las obras y las características de los proyectos DS27, identificando aquellos factores que podrían explicar las diferencias observadas en los plazos de ejecución.

**Detección de Valores Atípicos**

Se utilizará estadísticas descriptivas y el rango intercuartílico (IQR) para detectar valores atípicos en variables cuantitativas críticas como `Duracion_Obra_Dias` y `Total_Asignacion_UF`.

```python
# Resumen estadístico básico de las variables clave para identificar anomalías extremas
display(df[['Duracion_Obra_Dias', 'Total_Asignacion_UF', 'Numero_Unidades', 'Beneficiarios']].describe())

# Método del Rango Intercuartílico (IQR) para identificar potenciales valores atípicos en la variable objetivo (Duración)
Q1 = df['Duracion_Obra_Dias'].quantile(0.25)
Q3 = df['Duracion_Obra_Dias'].quantile(0.75)
IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

outliers_duracion = df[(df['Duracion_Obra_Dias'] < limite_inferior) | (df['Duracion_Obra_Dias'] > limite_superior)]
print(f"Límite inferior para outliers (Duración): {limite_inferior}")
print(f"Límite superior para outliers (Duración): {limite_superior}")
print(f"Cantidad de valores atípicos identificados en la duración de obras: {len(outliers_duracion)}")
if len(outliers_duracion) > 0:
    display(outliers_duracion[['Codigo_SERVIU', 'Duracion_Obra_Dias', 'Total_Asignacion_UF']])
```

Para identificar posibles valores atípicos en la variable Duracion_Obra_Dias, se aplicó el método del rango intercuartílico (IQR) que permite detectar valores atípicos. Los resultados muestran que la duración promedio de las obras es de 110 días, con valores que oscilan entre 39 y 358 días. Mediante este procedimiento se identificaron 6 proyectos con duraciones superiores al límite esperado de 218,5 días, los cuales fueron clasificados como potenciales valores atípicos. Estos casos corresponden a proyectos que presentan tiempos de ejecución considerablemente mayores que el resto de la muestra.

Los valores atípicos identificados no fueron eliminados de la base de datos, ya que no corresponden necesariamente a errores de registro. Dado que representan proyectos reales con duraciones excepcionalmente altas, se decidió conservarlos para el análisis, puesto que podrían reflejar condiciones particulares de ejecución y aportar información relevante para el estudio de los factores que influyen en la duración de las obras.

##**Etapa 5. Modelo analítico aplicado**



```python
import statsmodels.api as sm
import pandas as pd

# Crear variables dummies para Capitulo_DS27 con tipo float/int
dummies_capitulo = pd.get_dummies(df['Capitulo_DS27'], prefix='Capitulo', drop_first=True, dtype=float)

# Unir las variables independientes
X = df[['Total_Asignacion_UF', 'Numero_Unidades', 'Obra_Paralizada']].copy()
X = pd.concat([X, dummies_capitulo], axis=1)

# Variable dependiente
y = df['Duracion_Obra_Dias']

# Agregar constante al modelo para intercepto
X = sm.add_constant(X)

# Ajustar el modelo de regresión por Mínimos Cuadrados Ordinarios (OLS)
modelo = sm.OLS(y, X).fit()

# Mostrar el resumen estadístico del modelo
print(modelo.summary())
```

Se aplicó un modelo de regresión lineal múltiple, ya que el objetivo de la investigación es identificar qué factores influyen en la duración de ejecución de las obras. Esta técnica permite evaluar simultáneamente el efecto de varias variables explicativas sobre una variable dependiente cuantitativa, en este caso la duración de la obra medida en días.

**Variable dependiente (Y):**

Duracion_Obra_Dias

**Variables independientes (X):**

Total_Asignacion_UF

Numero_Unidades

Obra_Paralizada

Capitulo_DS27

**Resultados:** El modelo logra explicar aproximadamente el 55,7% de la variabilidad observada en la duración de las obras, lo que indica una capacidad explicativa moderada. En consecuencia, existen otros factores no incorporados en el modelo que también podrían influir en los plazos de ejecución.

R² = 0,557

Adj. R² = 0,534

Los resultados de la regresión lineal múltiple indican que la condición de paralización es el principal factor asociado a la duración de ejecución de las obras DS27. Manteniendo constantes las demás variables del modelo, los proyectos paralizados presentan en promedio aproximadamente 131 días adicionales de duración respecto de aquellos que no registran paralizaciones.

En contraste, el monto de asignación no mostró una asociación estadísticamente significativa con la duración de las obras (p = 0,657), por lo que, no existe evidencia suficiente para afirmar que influye en los tiempos de ejecución dentro de esta muestra.

El número de unidades tampoco presentó un efecto estadísticamente significativo sobre la duración de las obras (p = 0,678).

Aunque algunos capítulos presentan coeficientes positivos, las diferencias observadas no son estadísticamente significativas, por lo que no puede afirmarse que la tipología del proyecto influya de manera concluyente sobre la duración de las obras en esta muestra

En consecuencia, la investigación sugiere que los factores asociados al estado de ejecución del proyecto tienen una mayor influencia sobre la duración de las obras que sus características financieras o de tamaño.

La condición de paralización fue el factor con mayor influencia sobre la duración de las obras.

Los proyectos paralizados presentan tiempos de ejecución considerablemente superiores a los proyectos sin paralización.

Existen diferencias en la duración promedio entre los distintos capítulos del programa DS27.

El monto asignado presenta una relación débil con la duración de las obras.

Factores asociados a la gestión de la ejecución tienen una mayor incidencia en los plazos que las variables financieras analizadas.

Aunque inicialmente se esperaba que el monto de inversión tuviera una influencia importante sobre la duración de las obras, los resultados muestran que proyectos con mayores recursos no necesariamente presentan mayores tiempos de ejecución. Esto sugiere que las características técnicas de la intervención y las condiciones de ejecución pueden ser más determinantes que el nivel de financiamiento por sí solo.

**Limitaciones**

La variable Capítulo DS27 agrupa distintos tipos de intervenciones dentro de una misma categoría. En la práctica, proyectos de mejoramiento, ampliación y equipamiento pueden presentar requerimientos constructivos muy diferentes, lo que podría afectar la precisión del análisis.

El monto asignado refleja el presupuesto total del proyecto, pero no necesariamente su complejidad constructiva. Existen intervenciones de alto costo que pueden ejecutarse en plazos reducidos y obras de menor inversión que requieren más tiempo de ejecución.

**Recomendaciones**

Realizar un seguimiento preventivo a proyectos que presenten riesgos de paralización, especialmente aquellos vinculados a dificultades financieras de las empresas constructoras.

Profundizar el análisis por tipo de intervención dentro de cada capítulo DS27, considerando que las exigencias técnicas de una ampliación, un mejoramiento o una obra comunitaria pueden generar diferencias relevantes en los tiempos de ejecución.
Utilizar los resultados como apoyo para la programación y monitoreo de proyectos, priorizando la identificación temprana de riesgos que puedan afectar la continuidad de las obras.

La duración de las obras DS27 se encuentra más influenciada por factores asociados a la continuidad y gestión de la ejecución de los proyectos que por el monto de recursos asignados, lo que refuerza la importancia de monitorear oportunamente los riesgos que pueden derivar en paralizaciones de obra.
